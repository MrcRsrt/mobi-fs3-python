**<a id="#">Task 4.1.2.</a>** Implement horizontal and vertical 3&times;3 Prewitt filters:

```python
def prewitt_h(img_in):
    result = np.zeros(img_in.shape)
    vec = 1/6*np.array([-1, 0, 1])
    n = (3-1)//2 
    for p in numpy.ndindex(img_in.shape):
        neighborhood_sum = 0
        if p[0] - n < 0:
            continue
        elif p[0] + n >= img.shape[0]:
            continue
        elif p[1] - n < 0:
            continue
        elif p[1] + n >= img.shape[1]:
            continue
        result[p[0]][p[1]] = np.sum(img_in[(p[0]-1):(p[0]+2), (p[1]-1):(p[1]+2)]*vec)
    return result
    

def prewitt_v(img_in):
    result = np.zeros(img_in.shape)
    vec = 1/6*np.array([-1, 0, 1])
    n = (3-1)//2 
    for p in numpy.ndindex(img_in.shape):
        neighborhood_sum = 0
        if p[0] - n < 0:
            continue
        elif p[0] + n >= img.shape[0]:
            continue
        elif p[1] - n < 0:
            continue
        elif p[1] + n >= img.shape[1]:
            continue
        result[p[0]][p[1]] = np.sum(np.transpose(img_in[(p[0]-1):(p[0]+2), (p[1]-1):(p[1]+2)])*vec)
    return result
```
